```python
# Step 1: Data Preparation
# Define the data for alliances, seats, and votes
alliances = ["NDA", "INDIA", "Others"]
seats = [293, 234, 16]  # Seats won by NDA, INDIA, Others
votes = [235974144, 136759064, 13316039]  # Votes received by NDA, INDIA, Others

# Step 2: Calculate Total Seats and Votes
total_seats = sum(seats)
total_votes = sum(votes)

# Step 3: Determine the Overall Winner
overall_winner = alliances[seats.index(max(seats))]
print(f"The overall election winner is: {overall_winner} with {max(seats)} seats.")

# Step 4: Calculate Vote Share Percentages
vote_share = [(votes[i] / total_votes) * 100 for i in range(len(alliances))]
for i in range(len(alliances)):
    print(f"Vote share for {alliances[i]}: {vote_share[i]:.2f}%")

# Step 5: Identify Close Contests (States/Regions with Small Seat Margins)
# Example: Define state-wise seat distribution
states = {
    "Uttar Pradesh": {"NDA": 36, "INDIA": 43, "Others": 1},
    "Maharashtra": {"NDA": 17, "INDIA": 30, "Others": 1},
    "West Bengal": {"NDA": 12, "INDIA": 30, "Others": 0},
    "Bihar": {"NDA": 30, "INDIA": 9, "Others": 1},
    "Tamil Nadu": {"NDA": 0, "INDIA": 39, "Others": 0},
}

# Identify states with close contests (margin < 5 seats)
close_contests = {}
for state, results in states.items():
    sorted_results = sorted(results.values(), reverse=True)
    margin = sorted_results[0] - sorted_results[1]
    if margin < 5:
        close_contests[state] = margin

print("\nStates with close contests (margin < 5 seats):")
for state, margin in close_contests.items():
    print(f"{state}: Margin = {margin} seats")

# Step 6: Regional Breakdown
# Example: Define regional seat distribution
regions = {
    "North India": {"NDA": 151, "INDIA": 83, "Others": 6},
    "West India": {"NDA": 78, "INDIA": 31, "Others": 2},
    "South India": {"NDA": 49, "INDIA": 77, "Others": 5},
    "East India": {"NDA": 72, "INDIA": 45, "Others": 1},
    "Northeast India": {"NDA": 16, "INDIA": 7, "Others": 2},
}

# Print regional seat distribution
print("\nRegional Seat Distribution:")
for region, results in regions.items():
    print(f"{region}: NDA = {results['NDA']}, INDIA = {results['INDIA']}, Others = {results['Others']}")
```
